**Additional Exercise #1: Library Book Management System**
Functionality: Develop a program that allows a library to manage books, including adding new books, borrowing, and returning them.

Input:

• The user enters the number of books to add.

• For each book, the user provides a title, author, and availability status.

• The user can borrow a book by entering its title.

• The user can return a borrowed book.

Calculations:

• Keep track of which books are available, and which are borrowed.

• Ensure that a book cannot be borrowed if it is checked out.

Output:

• Display the list of books with their status (available/borrowed).

• Confirm when a book is successfully borrowed or returned.

Handle Invalid Inputs:

• Reject borrowing a book that doesn’t exist or is already borrowed.

• Ensure book titles are entered as strings

```python
# EXERCISE 1: Library Book Management System (FIXED)

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author):
        book = Book(title, author)
        self.books.append(book)
        print(f"Book '{title}' by {author} added successfully.")

    def display_books(self):
        if not self.books:
            print("No books available in the library.")
            return
        print("\n--- Library Books ---")
        for book in self.books:
            status = "Available" if book.is_available else "Borrowed"
            print(f"Title: {book.title} | Author: {book.author} | Status: {status}")
        print("---------------------")

    def borrow_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if book.is_available:
                    book.is_available = False
                    print(f"You have successfully borrowed '{book.title}'.")
                    return
                else:
                    print(f"Sorry, '{book.title}' is already borrowed.")
                    return
        print(f"Book '{title}' does not exist in the library.")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if not book.is_available:
                    book.is_available = True
                    print(f"You have successfully returned '{book.title}'.")
                    return
                else:
                    print(f"'{book.title}' was not borrowed.")
                    return
        print(f"Book '{title}' does not exist in the library.")


# --- Main Program ---
library = Library()

print("=== Add Books to Library ===")
print("Type 'done' as the title when you are finished adding books.\n")

while True:
    title = input("Enter book title (or 'done' to finish): ")
    if title.lower() == "done":
        break
    author = input(f"Enter author of '{title}': ")
    library.add_book(title, author)

while True:
    print("\n--- Library Menu ---")
    print("[1] Display Books")
    print("[2] Borrow a Book")
    print("[3] Return a Book")
    print("[4] Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
        library.display_books()
    elif choice == "2":
        title = input("Enter the title of the book to borrow: ")
        library.borrow_book(title)
    elif choice == "3":
        title = input("Enter the title of the book to return: ")
        library.return_book(title)
    elif choice == "4":
        print("Exiting the Library System. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter 1, 2, 3, or 4.")
```

**Additional Exercise #2: Employee Payroll System**
Functionality: Create a payroll system that calculates the monthly salary of employees based on their hourly rate and hours worked.

Input:

• The user enters the number of employees.

• For each employee, the user provides a name, hourly wage, and total hours worked.

Calculations:

• Compute the monthly salary as:

Salary = Hourly Rate × Hours Worked

• If an employee works over 160 hours, calculate overtime pay at 1.5× the hourly rate for extra hours.

Output:

• Display each employee’s name, total hours worked, base salary, and overtime pay if applicable.

Handle Invalid Inputs:

• Reject negative hours or wages.

• Ensure at least one employee is entered.

```python
# EXERCISE 2: Employee Payroll System

class Employee:
    def __init__(self, name, hourly_wage, hours_worked):
        self.name = name
        self.hourly_wage = hourly_wage
        self.hours_worked = hours_worked

    def compute_salary(self):
        if self.hours_worked > 160:
            regular_pay = self.hourly_wage * 160
            overtime_hours = self.hours_worked - 160
            overtime_pay = overtime_hours * self.hourly_wage * 1.5
        else:
            regular_pay = self.hourly_wage * self.hours_worked
            overtime_pay = 0
        total_salary = regular_pay + overtime_pay
        return regular_pay, overtime_pay, total_salary

    def display_payroll(self):
        regular_pay, overtime_pay, total_salary = self.compute_salary()
        print(f"\nEmployee   : {self.name}")
        print(f"Hours Worked: {self.hours_worked} hrs")
        print(f"Base Salary : ${regular_pay:.2f}")
        if overtime_pay > 0:
            print(f"Overtime Pay: ${overtime_pay:.2f}")
        print(f"Total Salary: ${total_salary:.2f}")
        print("-" * 30)


# --- Main Program ---
while True:
    try:
        num_employees = int(input("Enter the number of employees: "))
        if num_employees < 1:
            print("Please enter at least one employee.")
            continue
        break
    except ValueError:
        print("Invalid input. Please enter a whole number.")

employees = []
for i in range(num_employees):
    print(f"\n--- Employee {i + 1} ---")
    name = input("Enter employee name: ")

    while True:
        try:
            hourly_wage = float(input("Enter hourly wage: $"))
            if hourly_wage < 0:
                print("Hourly wage cannot be negative.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    while True:
        try:
            hours_worked = float(input("Enter total hours worked: "))
            if hours_worked < 0:
                print("Hours worked cannot be negative.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    emp = Employee(name, hourly_wage, hours_worked)
    employees.append(emp)

print("\n===== PAYROLL REPORT =====")
for emp in employees:
    emp.display_payroll()
```

**Additional Exercise #3: Online Shopping Cart System**

Functionality: Develop a shopping cart system where users can add products, remove them, and view the total price.

Input:

• The user selects products to add from a predefined list.

• Each product has a name, price, and stock quantity.

• The user can remove items from the cart before checkout.

Calculations:

• Keep track of total cost as items are added/removed.

• Ensure stock availability before adding a product.

Output:

• Display the cart’s content with the total price.

• Confirm when an item is successfully added or removed.

Handle Invalid Inputs:

• Prevent adding more items than available stock.

• Reject invalid product selections.

```python
# EXERCISE 3: Online Shopping Cart System

class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock


class ShoppingCart:
    def __init__(self):
        self.cart = {}

    def add_item(self, product, quantity=1):
        if product.stock <= 0:
            print(f"Sorry, '{product.name}' is out of stock.")
            return
        if quantity > product.stock:
            print(f"Only {product.stock} unit(s) of '{product.name}' available.")
            return
        if product.name in self.cart:
            self.cart[product.name]["quantity"] += quantity
        else:
            self.cart[product.name] = {"product": product, "quantity": quantity}
        product.stock -= quantity
        print(f"'{product.name}' (x{quantity}) added to cart.")

    def remove_item(self, product_name):
        if product_name in self.cart:
            quantity = self.cart[product_name]["quantity"]
            product = self.cart[product_name]["product"]
            product.stock += quantity
            del self.cart[product_name]
            print(f"'{product_name}' removed from cart.")
        else:
            print(f"'{product_name}' is not in your cart.")

    def view_cart(self):
        if not self.cart:
            print("Your cart is empty.")
            return
        print("\n--- Your Cart ---")
        total = 0
        for name, details in self.cart.items():
            subtotal = details["product"].price * details["quantity"]
            total += subtotal
            print(f"{name} x{details['quantity']} - ${subtotal:.2f}")
        print(f"Total: ${total:.2f}")
        print("-----------------")

    def get_total(self):
        return sum(
            details["product"].price * details["quantity"]
            for details in self.cart.values()
        )


# --- Predefined Product List ---
products = [
    Product("Laptop", 999.99, 5),
    Product("Smartphone", 699.99, 10),
    Product("Headphones", 149.99, 8),
    Product("Keyboard", 79.99, 15),
    Product("Mouse", 49.99, 20),
]

cart = ShoppingCart()

while True:
    print("\n--- Online Shop Menu ---")
    print("[1] View Products")
    print("[2] Add Item to Cart")
    print("[3] Remove Item from Cart")
    print("[4] View Cart")
    print("[5] Checkout")
    print("[6] Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
        print("\n--- Available Products ---")
        for i, product in enumerate(products, 1):
            print(f"[{i}] {product.name} - ${product.price:.2f} | Stock: {product.stock}")

    elif choice == "2":
        print("\n--- Available Products ---")
        for i, product in enumerate(products, 1):
            print(f"[{i}] {product.name} - ${product.price:.2f} | Stock: {product.stock}")
        try:
            selection = int(input("Enter product number to add: "))
            if 1 <= selection <= len(products):
                selected = products[selection - 1]
                qty = int(input(f"Enter quantity for '{selected.name}': "))
                if qty <= 0:
                    print("Quantity must be at least 1.")
                else:
                    cart.add_item(selected, qty)
            else:
                print("Invalid product selection.")
        except ValueError:
            print("Invalid input. Please enter a number.")

    elif choice == "3":
        cart.view_cart()
        item_name = input("Enter the product name to remove: ")
        cart.remove_item(item_name)

    elif choice == "4":
        cart.view_cart()

    elif choice == "5":
        cart.view_cart()
        if cart.cart:
            print(f"Thank you for your purchase! Total paid: ${cart.get_total():.2f}")
            cart.cart.clear()
        else:
            print("Your cart is empty. Nothing to checkout.")

    elif choice == "6":
        print("Thank you for visiting! Goodbye!")
        break

    else:
        print("Invalid choice. Please enter a number from 1 to 6.")
```
